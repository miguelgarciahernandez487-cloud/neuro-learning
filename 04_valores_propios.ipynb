## Ejercicio 6: Valores y vectores propios
Calculo los valores y vectores propios de una matriz con `np.linalg.eig` y verifico la definición A v = λ v.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
M = np.array([[3,1],[0,2]])
valores, vectores = np.linalg.eig(M)
print("Valores propios: ", valores)
for i in range(vectores.shape[1]):
    print("Vector propio no.", i, ":", vectores[:,i])
```

```python
for i in range(len(valores)):
    v = vectores[:,i]
    lam = valores[i]
    print(f"Par {i}: lambda = {lam.real:.2f} ->", np.allclose(M@v, lam*v))
```

```python
v1 = vectores[:,0].real
v2 = vectores[:,1].real
Mv1 = M@v1
Mv2 = M@v2

plt.quiver(0,0,Mv1[0], Mv1[1],angles='xy', scale_units='xy', scale=1, color='crimson', alpha=0.6, width=0.012, label='M @ v1')
plt.quiver(0,0,v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', width=0.008, label='v1')
plt.quiver(0,0,Mv2[0], Mv2[1], angles='xy', scale_units='xy', scale=1, color='orange', alpha=0.6, width=0.012, label='M @ v2')
plt.quiver(0,0,v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='green', width=0.008, label='v2')
plt.gca().set_aspect("equal", adjustable="box")
plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.grid()
plt.legend()
plt.title("Vectores propios de M: M @ v queda sobre la misma línea que v")
plt.show()
```

```python
#Determinante
print(np.linalg.det(M), np.prod(valores).real) 
```

```python
v1 = vectores[:,0].real
v2 = vectores[:,1].real
a=np.radians(45)
R = np.array([[np.cos(a), -np.sin(a)], [np.sin(a), np.cos(a)]])
MvT1 = R@v1
MvT2 = R@v2
plt.title("R @ v: la rotación saca a los vectores de su línea")
plt.quiver(0,0,MvT1[0], MvT1[1],angles='xy', scale_units='xy', scale=1, color='crimson', alpha=0.6, width=0.012, label='R @ v1')
plt.quiver(0,0,v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='blue', width=0.008, label='v1')
plt.quiver(0,0,MvT2[0], MvT2[1], angles='xy', scale_units='xy', scale=1, color='orange', alpha=0.6, width=0.012, label='R @ v2')
plt.quiver(0,0,v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='green', width=0.008, label='v2')
plt.gca().set_aspect("equal", adjustable="box")
plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.grid()
plt.legend()
plt.show()
```

```python
valoresR, vectoresR = np.linalg.eig(R)
print(valoresR)
print(vectoresR)
```

**Conclusión:** Calculé los valores y vectores propios de M = [[3,1],[0,2]] con `np.linalg.eig`. Los valores propios son 3 y 2, y verifiqué para ambos pares que M @ v = λ v. En la gráfica, M @ v queda sobre la misma línea que v, estirado por el valor propio. En cambio, la rotación de 45° da valores propios complejos (0.707 ± 0.707j): ningún vector real conserva su dirección al girar el plano. También comprobé que el determinante de M (6) es el producto de sus valores propios (3 · 2).
