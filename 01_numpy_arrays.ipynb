```python
#Ejercicio 1
#Con bucle for
import numpy as np
m1 = np.arange(0,100)
for j in range(0,len(m1)):
    if m1[j] % 7 == 0:
        m1[j+1] = 1
    else:
        m1[j] = 0
np.nonzero(m1)
#Sin bucle for
m1 = np.arange(0,100)
mask = (m1 % 7 == 0)
extract_from_m1 = m1[mask]
```

```python
#Ejercicio 2
rng = np.random.default_rng()
a = rng.integers(0,100,25).reshape(5,5).astype(float)
b = a.copy()
#Con for
for j in range(a.shape[1]):
    p = a[:,j].mean()
    for i in range(a.shape[0]):
        a[i,j] = a[i,j] - p
#Sin for
b = b - b.mean(axis=0)
print(np.allclose(b.mean(axis=0), 0))
print(np.allclose(a.mean(axis=0), 0))
```

```python
#Ejercicio 3
rng = np.random.default_rng()
a = rng.integers(0,100,25).reshape(5,5).astype(float)
a = a - a.mean(axis=1, keepdims = True)
a = a / a.std(axis=1, keepdims = True)
print(np.allclose(a.mean(axis=1), 0))
print(np.allclose(a.std(axis=1), 1))
```
