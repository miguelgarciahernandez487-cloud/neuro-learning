```python
import numpy as np
import matplotlib.pyplot as plt
d = 5
f = 10
fs = 250
t = np.arange(0,d, 1/fs)
onda_seno_10 = np.sin(2*np.pi*f*t + 0)
onda_seno_25 = np.sin(2*np.pi*25*t + 0)
ruido = np.random.normal(loc=0.0, scale=0.4, size=len(t)) #loc -> media, scale -> desv_est
senal = onda_seno_10 + onda_seno_25 + ruido
tf = np.fft.rfft(senal)
fq = np.fft.rfftfreq(len(senal), d=1/fs)
magnitud = abs(tf)
idx_picos = np.argsort(magnitud)[-2:]

plt.subplot(2,1,1)
plt.plot(t,senal, color="blue", linestyle="-",label="Señal con ruido")
plt.plot(t,onda_seno_10, color="red", linestyle="-", label="Señal_10")
plt.plot(t,onda_seno_25, color="green", linestyle="-", label="Señal_25")
plt.legend(loc="best")
plt.title("Señal")
plt.xlabel("tiempo (s)")
plt.ylabel("Amplitud")

plt.subplot(2,1,2)
plt.plot(fq, magnitud, color="green")
for i in idx_picos:
    plt.axvline(fq[i], color="red", linestyle="--")
plt.xlim(0, 60)
plt.title("Espectro")
plt.xlabel("Frecuencia (Hz)")
plt.ylabel("Magnitud")
plt.show()
print("picos más grandes: ",fq[idx_picos])
```
